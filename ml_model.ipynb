```python
import pandas as pd
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
data=pd.read_excel("iris_data.xlsx")
a=data.iloc[:,:-1]
data.head()
```

```python
b=data.iloc[:,-1]
b
```

```python
model=RandomForestClassifier()
model.fit(a,b)
predition=model.predict([[5.5,3,3.3,1]])
print(predition)
```

```python
dataset=pd.read_excel("iris_test.xlsx")
dataset
```

```python
x_test=dataset.iloc[:,:-1]
x_test
```

```python
y_test=dataset.iloc[:,-1]
y_test
```

```python
predi=model.predict(x_test)
predi
```

```python
y_test.values
```

```python
actual=y_test.values
```

```python
count=0
for i in range(0,len(actual)):
    if actual[i]==predi[i]:
        count=count+1
print((count*100)/len(actual))
```
